```python
# Requires: pandas, numpy, faker
!pip install faker
import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta

fake = Faker()
Faker.seed(42)
np.random.seed(42)
random.seed(42)

N = 3000

cities = ["Karachi","Lahore","Islamabad","Rawalpindi","Faisalabad","Multan","Peshawar","Quetta"]
city_weights = [0.30,0.22,0.12,0.10,0.08,0.06,0.07,0.05]  # skewed to Karachi/Lahore
store_types = ["Supermarket","Grocery","Pharmacy","Online","Kiosk"]
store_weights = [0.35,0.30,0.15,0.12,0.08]
product_categories = ["Beverages","Snacks","Dairy","PersonalCare","Household","Electronics","Apparel"]
prod_weights = [0.20,0.20,0.15,0.15,0.15,0.08,0.07]
payment_methods = ["Cash","Card","MobileWallet","E-Transfer"]
customer_types = ["Regular","New","Corporate"]

# create product catalog (about 120 products across categories)
catalog = []
prod_id = 1000
for cat in product_categories:
    n = 20 if cat not in ("Electronics","Apparel") else 10
    for i in range(n):
        prod_code = f"P{prod_id}"
        name = f"{cat[:3].capitalize()} Product {i+1}"
        # base price by category (PKR)
        base = {
            "Beverages": 80,
            "Snacks": 60,
            "Dairy": 150,
            "PersonalCare": 250,
            "Household": 300,
            "Electronics": 3500,
            "Apparel": 1200
        }[cat]
        # vary base with multiplier
        price = base * np.random.lognormal(mean=0.0, sigma=0.6)
        catalog.append((prod_code, name, cat, round(price,2)))
        prod_id += 1

# helper formatters to introduce messy strings
def messy_price(p):
    fmt = random.choices(
        ["Rs. {v}", "PKR {v}", "{v} PKR", "{v}", "Rs {v}", "PKR{v}"],
        weights=[0.25,0.20,0.15,0.15,0.15,0.10],
        k=1
    )[0]
    v = f"{p:,.2f}" if random.random()<0.5 else f"{int(round(p)):,}"
    s = fmt.format(v=v)
    # random spaces and occasional trailing text
    if random.random()<0.08: s = " " + s + " "
    if random.random()<0.03: s = s + " only"
    return s

def messy_percent(p):
    # p as fraction like 0.05
    r = random.random()
    if r < 0.7:
        return f"{int(p*100)}%"
    elif r < 0.9:
        return f"{p:.2f}"
    else:
        return f"{p*100:.1f} %"

def messy_quantity(q):
    if random.random()<0.12:
        return f"{q} pcs"
    elif random.random()<0.05:
        return f"{q} units"
    else:
        return str(q)

rows = []
start_date = datetime(2024,1,1)
for i in range(N):
    invoice_id = f"INV2024{start_date.month:02d}{start_date.day:02d}_{i+1:04d}"
    # choose date randomly across 2024
    d = start_date + timedelta(days=int(np.random.uniform(0,365)))
    # 3% malformed date
    if random.random()<0.03:
        date = random.choice([d.strftime("%Y/%m/%d"), d.strftime("%b %d %Y")])
    else:
        date = d.strftime("%Y-%m-%d")
    city = random.choices(cities, weights=city_weights, k=1)[0]
    store = random.choices(store_types, weights=store_weights, k=1)[0]
    prod = random.choices(catalog, weights=[prod_weights[product_categories.index(x[2])] for x in catalog], k=1)[0]
    product_id, product_name, product_category, base_price = prod
    # make unit price: base_price * lognormal noise, produce right skew
    unit_price_num = max(10, np.random.lognormal(mean=np.log(base_price+1), sigma=0.9))
    # occasional electronics very high
    if product_category=="Electronics" and random.random()<0.1:
        unit_price_num *= random.uniform(1.0,3.0)
    unit_price_num = round(unit_price_num,2)
    unit_price = messy_price(unit_price_num)
    # quantity skew: mostly 1-3
    if random.random()<0.90:
        quantity_num = int(np.random.choice([1,2,3], p=[0.6,0.3,0.1]))
    else:
        quantity_num = int(np.random.randint(4,51))
    quantity = messy_quantity(quantity_num)
    # discount: mostly 0, some 5-20%
    if random.random()<0.7:
        disc = 0.0
    else:
        disc = random.choice([0.05,0.10,0.15,0.20])
    discount_pct = messy_percent(disc)
    # tax: many have 17%, some 0, some missing
    t = random.choices([0.17,0.0,0.0], weights=[0.5,0.45,0.05])[0]
    tax_pct = messy_percent(t)
    # compute amounts
    total_calc = round(unit_price_num * quantity_num,2)
    # sometimes total_amount provided but formatted or incorrect
    if random.random()<0.95:
        total_amount = messy_price(total_calc)
    else:
        # incorrect or missing
        if random.random()<0.5:
            total_amount = "NA"
        else:
            wrong = round(total_calc * random.uniform(0.6,1.4),2)
            total_amount = messy_price(wrong)
    # net_amount: apply discount and tax, but some rows missing or incorrect
    net_calc = round((total_calc * (1 - disc)) * (1 + t),2)
    if random.random()<0.9:
        # sometimes as numeric, sometimes formatted
        if random.random()<0.6:
            net_amount = messy_price(net_calc)
        else:
            net_amount = str(net_calc)
    else:
        net_amount = ""  # missing
    # payment, customer
    payment = random.choice(payment_methods)
    customer = random.choice(customer_types + [""])  # some missing
    # returned variations
    ret = random.choices(["No","Yes","N","Y","0","1",""], weights=[0.82,0.03,0.05,0.01,0.05,0.02,0.02], k=1)[0]
    notes = random.choice(["","promo applied","bulk order","damaged packaging","returned","NA","null"])
    # introduce occasional literal 'nan' strings
    if random.random()<0.03:
        unit_price = "nan"
    if random.random()<0.03:
        discount_pct = "nan"
    if random.random()<0.02:
        tax_pct = "nan"
    # compile row
    row = {
        "invoice_id": invoice_id,
        "date": date,
        "store_city": city,
        "store_type": store,
        "product_category": product_category,
        "product_id": product_id,
        "product_name": product_name,
        "unit_price": unit_price,
        "quantity": quantity,
        "total_amount": total_amount,
        "discount_pct": discount_pct,
        "customer_type": customer,
        "payment_method": payment,
        "returned": ret,
        "tax_pct": tax_pct,
        "net_amount": net_amount,
        "notes": notes
    }
    rows.append(row)

df = pd.DataFrame(rows)

# inject further missingness and typos
for col in ["unit_price","total_amount","net_amount"]:
    df.loc[df.sample(frac=0.03, random_state=1).index, col] = df[col].sample(frac=0.03).astype(str).map(lambda x: random.choice(["","NA","na","nan"]) )

# some payment_method typos
df.loc[df.sample(frac=0.02, random_state=2).index, "payment_method"] = df["payment_method"].sample(frac=0.02).map(lambda x: x.lower() if pd.notna(x) else x)
df.loc[df.sample(frac=0.01, random_state=3).index, "payment_method"] = df["payment_method"].sample(frac=0.01).map(lambda x: x + " " if pd.notna(x) else x)

# write CSV
df.to_csv("sales_pakistan_3000.csv", index=False)
print("Wrote sales_pakistan_3000.csv with", len(df), "rows")
```
